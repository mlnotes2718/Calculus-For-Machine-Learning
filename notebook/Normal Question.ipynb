## Apply Normal Equation for Intercept

**Cost Function**
$$J(w,b) = \frac {1}{2m} \sum \limits_{i=1}^m \left( f_{w,b}(x) - y \right)^2$$

**Derivative in Respect to Intercept**

$$J(w,b) = \frac {1}{2m} \sum \limits_{i=1}^m \left( (b + wx_i) - y_i \right)^2$$
$$J'(b) = \frac {1}{2m} \cdot 2 \cdot \sum \limits_{i=1}^m ((b + wx_i) -y_i)$$
$$J'(b) = \frac {1}{m} \cdot \sum \limits_{i=1}^m ((b + wx_i) -y_i) \cdot 1$$
$$0 = \frac {1}{m} \cdot \sum \limits_{i=1}^m ((b + wx_i) -y_i)$$

Multiply both side by m
$$0 \cdot m = \frac {1}{m} \cdot m \sum \limits_{i=1}^m ((b + wx_i) -y_i)$$
$$0 = 1 \cdot \sum \limits_{i=1}^m ((b + wx_i) -y_i)$$
$$0 = \sum \limits_{i=1}^m (b + wx_i) - y_i$$
$$0 = \sum \limits_{i=1}^m b + \sum \limits_{i=1}^m wx_i - \sum \limits_{i=1}^m y_i$$

$\sum \limits_{i=1}^m b = m \cdot b$ since b remain the same for all the $m$ lines
$$ \sum \limits_{i=1}^m y_i = mb + \sum \limits_{i=1}^m wx_1$$
$$ \sum \limits_{i=1}^m y_i - w \cdot \sum \limits_{i=1}^m x_i = mb$$
$$ mb = \sum \limits_{i=1}^m y_i - w \cdot \sum \limits_{i=1}^m x_i$$
$$ b = \frac {\sum \limits_{i=1}^m y_i}{m} - w \cdot \frac {\sum \limits_{i=1}^m x_i}{m}$$

$$ intercept = \frac {\sum \limits_{i=1}^m y_i}{m} - w \cdot \frac {\sum \limits_{i=1}^m x_i}{m}$$
$$$$
$$intercept = \text{(mean of y) } - \text{slope } \cdot \text{(mean of x)}$$

**Note: "Removing m" vs "Introducing m" — they're not the same m**

When solving $0 = \dfrac{1}{m}\sum_{i=1}^m(b+wx_i-y_i)$ for $b$, you'll see $m$ disappear and then reappear. These are two *unrelated* events that happen to use the same letter.

**1. Removing $m$ — clearing a division**
A fraction equals zero only when its numerator is zero (assuming $m \ne 0$). So:
$$\frac{1}{m}\sum(\dots) = 0 \quad\Longrightarrow\quad \sum(\dots) = 0$$
This $m$ is *cancelled* — it was already in the equation, and dividing/multiplying just removes it. Nothing new happens here.

**2. Introducing $m$ — counting repeated terms**
Now expand the sum:
$$\sum_{i=1}^m (b + wx_i - y_i) = \sum_{i=1}^m b + \sum_{i=1}^m wx_i - \sum_{i=1}^m y_i$$
$\sum_{i=1}^m b$ means "add $b$ to itself $m$ times" (since $b$ has no $i$ in it), which is just $mb$. This $m$ is *brand new* — it comes from counting how many terms are being summed, not from undoing step 1.

**Rule of thumb:** any time you see $\sum_{i=1}^m(\text{constant})$, it collapses to $m \times \text{constant}$. That's where a fresh $m$ always comes from — it's not the same $m$ you cleared earlier, even though it's the same symbol (same dataset size).

### Application of Intercept Formula

```python
def intercept(x,y, slope):
    result = y.mean() - (slope * x.mean())
    return result
```

## Apply Normal Equation for w

$$J(w,b) = \frac {1}{2m} \sum \limits_{i=1}^m \left( (b + wx_i) - y_i \right)^2$$

The partial derivative with respect to w is 
$$J'(w) = \frac {1}{m} \cdot \sum \limits_{i=1}^m ((b+ wx_i -y_i) \cdot x_i)$$

> Please refer to the notebook on derivation of gradient descent

$$J'(w) = \frac {1}{m} \cdot \sum \limits_{i=1}^m ((b+ wx_i -y_i) \cdot x_i)$$
$$J'(w) = \frac {1}{m} \cdot \sum \limits_{i=1}^m (b \cdot x_i + wx_i \cdot x_i - y_i \cdot x_i)$$
$$J'(w) = \frac {1}{m} \cdot \sum \limits_{i=1}^m (b \cdot x_i + wx_i^2 - y_i \cdot x_i)$$

Setting to $0$
$$0 = \frac {1}{m} \cdot \sum \limits_{i=1}^m (b \cdot x_i + wx_i^2 - y_i \cdot x_i)$$

multiply by $m$
$$0 \cdot m  = \frac {1}{m} \cdot m \cdot \sum \limits_{i=1}^m (b \cdot x_i + wx_i^2 - y_i \cdot x_i)$$
$$0 = \sum \limits_{i=1}^m (b \cdot x_i + wx_i^2 - y_i \cdot x_i)$$
$$0 = \sum \limits_{i=1}^m (bx_i + wx_i^2 - x_iy_i)$$
$$\sum \limits_{i=1}^m x_iy_i = \sum \limits_{i=1}^m bx_i + \sum \limits_{i=1}^mwx_i^2$$
$$\sum \limits_{i=1}^m bx_i + \sum \limits_{i=1}^mwx_i^2 = \sum \limits_{i=1}^m x_iy_i$$

$$b \sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$



Since $b = \bar y - w \bar x$

$$b \sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$
$$(\bar y - w \bar x )\sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$
$$\bar y\sum \limits_{i=1}^m x_i - w \bar x \sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$

$$\frac{\sum \limits_{i=1}^m y_i}{m}\sum \limits_{i=1}^m x_i - w \frac{\sum \limits_{i=1}^m x_i}{m} \sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$

$$\frac{\sum \limits_{i=1}^m y_i}{m}\sum \limits_{i=1}^m x_i - w \frac{(\sum \limits_{i=1}^m x_i)^2}{m} + w \sum \limits_{i=1}^m x_i^2 = \sum \limits_{i=1}^m x_iy_i$$

$$\frac{\sum \limits_{i=1}^m y_i}{m}\sum \limits_{i=1}^m x_i + w \sum \limits_{i=1}^m x_i^2 - w \frac{(\sum \limits_{i=1}^m x_i)^2}{m} = \sum \limits_{i=1}^m x_iy_i$$

$$\frac{(\sum \limits_{i=1}^m y_i) (\sum \limits_{i=1}^m x_i)}{m} + w \left ( \sum \limits_{i=1}^m x_i^2 - \frac{(\sum \limits_{i=1}^m x_i)^2}{m} \right ) = \sum \limits_{i=1}^m x_iy_i$$

$$w \left ( \sum \limits_{i=1}^m x_i^2 - \frac{(\sum \limits_{i=1}^m x_i)^2}{m} \right ) = \sum \limits_{i=1}^m x_iy_i - \frac{(\sum \limits_{i=1}^m y_i) (\sum \limits_{i=1}^m x_i)}{m}$$
