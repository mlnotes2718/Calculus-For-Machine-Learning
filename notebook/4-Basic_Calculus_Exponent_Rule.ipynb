Using what we have learn in book 3, we know that for a speed, distance and time function. If the distance vs time is ($f=t^2$), then the derivative is $2t$.

```python
t = np.linspace(-5, 5, 400)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(t, t**2)
axes[0].set_title(r"$f(t) = t^2$")
axes[0].set_xlabel("t")
axes[0].set_ylabel("f(t)")
axes[0].grid(True)

axes[1].plot(t, 2 * t, color="orange")
axes[1].set_title(r"$f'(t) = 2t$")
axes[1].set_xlabel("t")
axes[1].set_ylabel("f'(t)")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## Derivatives of Exponential Function

We can see that there is a pattern such that, if a function is:

$$f(x) = x^n$$

Then the derivatives are:

$$f'(x) = nx^{n-1}$$

The proper derivation of exponential rule is from Binomial Theorem.

# Deriving the Power Rule from the Binomial Theorem

**Goal:** show that $\dfrac{d}{dx}x^n = nx^{n-1}$ using the limit definition
of the derivative plus the binomial theorem (works cleanly for positive
integer $n$).

## 1. Start from the limit definition

$$f(x) = x^n \qquad\Rightarrow\qquad f'(x) = \lim_{h\to 0} \frac{(x+h)^n - x^n}{h}$$

## 2. Expand $(x+h)^n$ with the binomial theorem

$$(x+h)^n = \sum_{k=0}^{n} \binom{n}{k} x^{n-k}h^k$$

Written out, the first few terms:

$$(x+h)^n = x^n + \binom{n}{1}x^{n-1}h + \binom{n}{2}x^{n-2}h^2 + \cdots + h^n$$

## 3. Subtract $x^n$

The leading $x^n$ term cancels with the $-x^n$ from the difference quotient:

$$(x+h)^n - x^n = \binom{n}{1}x^{n-1}h + \binom{n}{2}x^{n-2}h^2 + \cdots + h^n$$

## 4. Divide by $h$

Every remaining term has at least one factor of $h$, so it divides out
cleanly:

$$\frac{(x+h)^n - x^n}{h} = \binom{n}{1}x^{n-1} + \binom{n}{2}x^{n-2}h + \cdots + h^{n-1}$$

## 5. Take the limit as $h \to 0$

Every term still carrying an $h$ (all except the first) vanishes:

$$f'(x) = \binom{n}{1}x^{n-1} = nx^{n-1}$$

since $\binom{n}{1} = n$.

## Key idea to remember

The binomial expansion is what generates the $h$-dependent "extra" terms
in $(x+h)^n$. Only the very first term ($\binom{n}{1}x^{n-1}h$) survives
division by $h$ *and* the limit $h\to 0$ without vanishing — every other
term still has a leftover $h$ that goes to zero. That surviving term is
exactly the power rule.

## Note on scope

This derivation works directly for **positive integers $n$** (where the
binomial theorem applies as-is). The rule also holds for negative,
fractional, and irrational $n$, but those cases need different tools
(e.g. the general binomial series, implicit differentiation, or the
definition via $e^{n\ln x}$) — worth exploring separately.
